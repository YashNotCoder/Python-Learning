```python
num = 5 
for i in range(0,num):
    for j in range(0,i+1):
     print("*" , sep = " " , end = " " )
    print()
```

```python
num = 5 
for i in range(num, 0, -1):
    for j in range(1,i+1):
     print("*" , sep = " " , end = " " )
    print()
```

```python
num = 5 
for i in range(1 , num+1):
    print(" " * (num - i), end = "")
    print("*" * (2 * i - 1))
 
```

```python
n = 5
for i in range(n):
   for j in range(i, n):
     print(' ', end='')
   for j in range(i+1):
     print('*', end='')
   print()
```

# List vs Tuple vs Set vs Dictionary

| Feature | List | Tuple | Set | Dictionary |
|---|---|---|---|---|
| **Syntax** | `[]` | `()` | `{}` | `{key: value}` |
| **Ordered** | ✅ Yes | ✅ Yes | ❌ No* | ✅ Yes** |
| **Mutable** | ✅ Yes | ❌ No | ✅ Yes | ✅ Yes |
| **Duplicates** | ✅ Allowed | ✅ Allowed | ❌ Not allowed | ❌ Keys cannot duplicate |
| **Indexing** | ✅ Yes | ✅ Yes | ❌ No | ❌ By index |
| **Stores** | Values | Values | Unique values | Key-value pairs |
| **Access example** | `my_list[0]` | `my_tuple[0]` | Not by index | `my_dict["name"]` |
| **Best used for** | Collection of items that may change | Fixed collection of items | Unique items | Data with labels/keys |
| **Example** | `["apple", "mango"]` | `("apple", "mango")` | `{"apple", "mango"}` | `{"fruit": "apple", "price": 100}` |

> **Note:** Sets are unordered collections for practical purposes. Dictionaries preserve insertion order in modern Python.

## Quick Memory Trick

- **List** → Ordered + Changeable + Duplicates
- **Tuple** → Ordered + Not Changeable
- **Set** → Unique values + No indexing
- **Dictionary** → Key → Value

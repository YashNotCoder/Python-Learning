# 'if-else' Condition

```python
age = 22
if age > 18 :
    print("NOt child")
else :
    print("Child")
```

# 'if-elif-else' Condition

```python
score = int(input("Enter Marks :  "))
if score >= 90 :
    print("A")
elif score >= 80 :
    print("B")
elif score >= 70 :
    print("C")
else :
    print("D")
```

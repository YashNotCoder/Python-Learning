#### FUNCTION WITH PARAMETERS

```python
def my_sum(a,b) :
    print(a + b)
```

```python
def my_sum2(a,b) :
    result = a + b 
    print("YOur SUM is:",result) 
```

## Return Statement
#### return is used to get the result as well as to preserve its datatype as in print function the data type becomes 'nonetype' , return is used to use the value later in the program

```python
def sub(a,b) :
    return a - b 
```

```python
solv = sub(10,6)
print(solv) 
type(solv)
```

## Pass Statement 
#### It is used to keep a function incomplete and can be modified later the next lines of code is executed , the function can be later completed

```python
def newfunction() :
   pass 

print("HEllo")
```

# 'if-else' Condition

```python
age = 22
if age > 18 :
    print("NOt child")
else :
    print("Child")
```

# 'if-elif-else' Condition

```python
score = int(input("Enter Marks :  "))
if score >= 90 :
    print("A")
elif score >= 80 :
    print("B")
elif score >= 70 :
    print("C")
else :
    print("D")
```

# Nested 'if-else' Condition

```python
num = int(input("Enter a number : "))
if num > 0 :
    if num % 2 == 0 :
        print("num is positive and even")
    else :
        print("num is positive and odd")
else :
    if num == 0 :
        print("num is 0")
    else :
        print("num is negative and not zero")
    
```

# Ternary operator

### Syntax :
#### value_if_true if condition else value_if_false 

```python
age = 22
status = "Adult" if age > 18 else "Child"
print(status)
```

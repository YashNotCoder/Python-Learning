# While Loop


### The while loop repeatedly executes a block of code as long as a given
### condition remains True. It checks the condition before each iteration.



```python
count = 0
while count < 4: # Condition
   print(count)
   count += 1
# Output: 0 1 2 3
```

# For Loop
#### The for loop in Python is used to iterate over a sequence (such as a list,
#### tuple, dictionary, set, or string) and execute a block of code for each
#### element in that sequence.


```python
language = 'Python'
for x in language:
  print(x)
```

```python
count = 'yes'
for x in count :
 print (x)
```

# Loop Control Statement

### Loop control statements allow you to alter the normal flow of a loop.
#### Python supports 3 clauses within loops:
##### • pass statement - learnt earlier used to leave code empty for writing it later
##### • break Statement
##### • continue Statement

## Loop Control - break Statement
#### break Statement: The break statement terminates the loop
#### entirely, exiting from it immediately.


```python
for i in range(5):
 if i == 3:
    break
print(i) # Output: 0 1 2
```

### Continue

```python
for i in range(5):
 if i == 3:
   continue  ##  skips the value and prints the rest
 print(i)
```

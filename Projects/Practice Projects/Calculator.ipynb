# Calculator

```python
print("WELCOME USER")

def add(num1,num2) :
    return num1 + num2

def sub(num1,num2) :
    return num1 - num2

def mul(num1,num2) :
    return num1 * num2

def div(num1,num2) :
    return num1 / num2

def avg(num1,num2) :
    return (num1 + num2) / 2

num1 = int(input("Enter numb1:" ))
num2 = int(input("Enter numb2:" ))
opr = input("Enter opr:")





if opr == '+' or opr == 'add' :
    print("Addition is" ,add(num1,num2))
elif opr == '-' or opr == 'sub' :
    print("Subtraction is", sub(num1,num2))
elif opr == '*' or opr == 'mul' :
    print("Multiplication is" ,mul(num1,num2))
elif opr == '/' or opr == 'div' :
    print("Division is", div(num1,num2))
elif opr == 'avg' :
    print("Average is",avg(num1,num2)) 
else :
    print("Enter valid operation")
     
    
```

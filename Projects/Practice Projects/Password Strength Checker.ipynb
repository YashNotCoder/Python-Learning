```python
import re 

def check_password_strength(password):
    if len(password) < 8:
        return "Weak : Password must be atleast 8 characters"

    if not any(char.isdigit() for char in password):
        return "Weak : Password must contain digit"

    if not any(char.isupper() for char in password):
        return "Weak : Password must conatin an uppercase alphabet"

    if not any(char.islower() for char in password):
        return "Weak : Password must conatin a lowercase alphabet"

    if not re.search(r'[!@#$%^&*(){}<>.?]' , password):
        return "Medium : password must contain a special character"

    return "Strong : Your password is secured !"



def password_checker() :
    print("WELCOME USER !!")
    while True: 
       password = input("Enter your password(or type 'exit to quit) : ")
        
       if password.lower() == 'exit' :
           print("Thank You for using")
           break 

       result = check_password_strength(password)
       print(result)


if __name__ == "__main__":
   password_checker()
```

```python
import re 

def check_pasword_strength(password):
    if len(password) < 8:
        return "Weak : Password must be atleast 8 characters"

    if not any(char.isdigit() for char in password):
        return "Weak : Password must contain digit"

    if not any(char.isupper() for char in password):
        return "Weak : Password must conatin an uppercase alphabet"

    if not any(char.islower() for char in password):
        return "Weak : Password must conatin a lowercase alphabet"
```

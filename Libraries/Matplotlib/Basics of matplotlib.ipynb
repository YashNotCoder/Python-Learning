```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5,6,7,8,9]
y = [10,20,30,40,50,60,70,80,90]
plt.plot(x,y)  # To plot a line chart
 
plt.show()    # To display the chart
```

```python
# Customise Chart
```

```python
plt.figure(figsize = (8,5))    #Figure size (horizontal , vertical)
plt.plot(x,y , color = 'Green' , marker = 'o' , linestyle = 'dashed' , markersize = 8) 
plt.title('Practicing plotting')
plt.xlabel("This is x axis")
plt.ylabel("This is y axis")
plt.show()
```

```python
list1 = [1,2,3]
print(list1)
arr1 = np.array(list1)
print(arr1)
```

```python
# Two dimensional array
```

```python
list1 = [1,2,3]
list2 = [4,5,6]

arr2 = np.array([list1, list2])
print(arr2)
```

```python
print("size" , arr2.size)
print("shape" , arr2.shape)
print("Dimension" , arr2.ndim)
print("dtype" , arr2.dtype)
```

```python
# Array initialization methods
```

```python
zero_arr = np.zeros((3,3))  #create an array full of zeros
print(zero_arr)
```

```python
one_arr = np.ones((3,3))   #create an array full of ones
print(one_arr)
```

```python
full_arr = np.full((3,3) , 7)   #create an array full of the number passed 
print(full_arr)
```

```python
ident_arr = np.eye(3)   #create an identity matrix
print(ident_arr)
```

```python
print(np.arange(1,10,1)) # creates and array of evenly spaced numbers (start,stop,step)
```

```python
print(np.linspace(1,10,2)) # creates an array of equally spaced number
```

```python
rand = np.random.rand(3,3) # Generates a array of random float values
print(rand)
```

```python
randint = np.random.randint(1,10 , (3,3))  #Generates array of integers (start,stop,dimension of array)
print(randint)
```
